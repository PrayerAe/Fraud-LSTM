```python
"""
Notebook 3: LSTM-Autoencoder Model Training
Deteksi Anomali Transaksi Real-Time Menggunakan LSTM-Autoencoder
"""

# ============================================================================
# CELL 1: Import Libraries
# ============================================================================
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import time
import warnings
warnings.filterwarnings('ignore')

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader, TensorDataset

# Set random seeds
np.random.seed(42)
torch.manual_seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed(42)

# Device configuration
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")
    print(f"Memory Allocated: {torch.cuda.memory_allocated(0) / 1024**2:.2f} MB")
    print(f"Memory Cached: {torch.cuda.memory_reserved(0) / 1024**2:.2f} MB")
```

```python
# ============================================================================
# CELL 2: Load Processed Data
# ============================================================================
print("\n" + "="*80)
print("LOADING PROCESSED DATA")
print("="*80)

# Load fraud dataset
X_train_fraud = np.load('outputs/data/X_train_fraud.npy')
y_train_fraud = np.load('outputs/data/y_train_fraud.npy')
X_val_fraud = np.load('outputs/data/X_val_fraud.npy')
y_val_fraud = np.load('outputs/data/y_val_fraud.npy')
X_test_fraud = np.load('outputs/data/X_test_fraud.npy')
y_test_fraud = np.load('outputs/data/y_test_fraud.npy')

print("FRAUD DATASET:")
print(f"  Train: {X_train_fraud.shape}")
print(f"  Val: {X_val_fraud.shape}")
print(f"  Test: {X_test_fraud.shape}")

# Load olist dataset
X_train_olist = np.load('outputs/data/X_train_olist.npy')
y_train_olist = np.load('outputs/data/y_train_olist.npy')
X_val_olist = np.load('outputs/data/X_val_olist.npy')
y_val_olist = np.load('outputs/data/y_val_olist.npy')
X_test_olist = np.load('outputs/data/X_test_olist.npy')
y_test_olist = np.load('outputs/data/y_test_olist.npy')

print("\nOLIST DATASET:")
print(f"  Train: {X_train_olist.shape}")
print(f"  Val: {X_val_olist.shape}")
print(f"  Test: {X_test_olist.shape}")

# Load configuration
config = np.load('outputs/data/config.npy', allow_pickle=True).item()
TIMESTEP = config['timestep']
N_FEATURES_FRAUD = config['n_features_fraud']
N_FEATURES_OLIST = config['n_features_olist']

print(f"\nConfiguration:")
print(f"  Timestep: {TIMESTEP}")
print(f"  Fraud features: {N_FEATURES_FRAUD}")
print(f"  Olist features: {N_FEATURES_OLIST}")
```

```python
# ============================================================================
# CELL 3: Define LSTM Autoencoder Architecture
# ============================================================================
print("\n" + "="*80)
print("DEFINING LSTM-AUTOENCODER ARCHITECTURE")
print("="*80)

class LSTMAutoencoder(nn.Module):
    def __init__(self, n_features, timestep, hidden_dim1=64, hidden_dim2=32, 
                 latent_dim=16, dropout=0.2):
        super(LSTMAutoencoder, self).__init__()
        
        self.n_features = n_features
        self.timestep = timestep
        self.hidden_dim1 = hidden_dim1
        self.hidden_dim2 = hidden_dim2
        self.latent_dim = latent_dim
        
        # Encoder
        self.encoder_lstm1 = nn.LSTM(
            input_size=n_features,
            hidden_size=hidden_dim1,
            num_layers=1,
            batch_first=True,
            dropout=0 if dropout == 0 else dropout
        )
        
        self.encoder_lstm2 = nn.LSTM(
            input_size=hidden_dim1,
            hidden_size=hidden_dim2,
            num_layers=1,
            batch_first=True,
            dropout=0 if dropout == 0 else dropout
        )
        
        # Latent space
        self.encoder_dense = nn.Linear(hidden_dim2, latent_dim)
        self.dropout = nn.Dropout(dropout)
        
        # Decoder
        self.decoder_dense = nn.Linear(latent_dim, hidden_dim2)
        
        self.decoder_lstm1 = nn.LSTM(
            input_size=hidden_dim2,
            hidden_size=hidden_dim2,
            num_layers=1,
            batch_first=True,
            dropout=0 if dropout == 0 else dropout
        )
        
        self.decoder_lstm2 = nn.LSTM(
            input_size=hidden_dim2,
            hidden_size=hidden_dim1,
            num_layers=1,
            batch_first=True,
            dropout=0 if dropout == 0 else dropout
        )
        
        # Output layer
        self.output_layer = nn.Linear(hidden_dim1, n_features)
        
    def forward(self, x):
        # x shape: (batch, timestep, features)
        
        # Encoding
        enc1, _ = self.encoder_lstm1(x)  # (batch, timestep, hidden_dim1)
        enc2, (hidden, cell) = self.encoder_lstm2(enc1)  # (batch, timestep, hidden_dim2)
        
        # Take the last output for encoding
        encoded = enc2[:, -1, :]  # (batch, hidden_dim2)
        
        # Latent representation
        latent = self.encoder_dense(encoded)  # (batch, latent_dim)
        latent = torch.relu(latent)
        latent = self.dropout(latent)
        
        # Decoding
        decoded = self.decoder_dense(latent)  # (batch, hidden_dim2)
        decoded = torch.relu(decoded)
        
        # Repeat for timesteps
        decoded = decoded.unsqueeze(1).repeat(1, self.timestep, 1)  # (batch, timestep, hidden_dim2)
        
        dec1, _ = self.decoder_lstm1(decoded)  # (batch, timestep, hidden_dim2)
        dec2, _ = self.decoder_lstm2(dec1)  # (batch, timestep, hidden_dim1)
        
        # Output reconstruction
        output = self.output_layer(dec2)  # (batch, timestep, features)
        
        return output
    
    def encode(self, x):
        """Get latent representation"""
        with torch.no_grad():
            enc1, _ = self.encoder_lstm1(x)
            enc2, _ = self.encoder_lstm2(enc1)
            encoded = enc2[:, -1, :]
            latent = self.encoder_dense(encoded)
            latent = torch.relu(latent)
        return latent

print("✓ LSTM-Autoencoder class defined!")
```

```python
# ============================================================================
# CELL 4: Training Function
# ============================================================================
print("\n" + "="*80)
print("DEFINING TRAINING FUNCTIONS")
print("="*80)

def train_autoencoder(model, train_loader, val_loader, criterion, optimizer, 
                      num_epochs, device, model_name='model'):
    """
    Train the autoencoder model
    """
    train_losses = []
    val_losses = []
    best_val_loss = float('inf')
    patience = 10
    patience_counter = 0
    
    print(f"\nTraining {model_name}...")
    print(f"{'Epoch':<10} {'Train Loss':<15} {'Val Loss':<15} {'Time':<10}")
    print("-" * 55)
    
    for epoch in range(num_epochs):
        start_time = time.time()
        
        # Training phase
        model.train()
        train_loss = 0.0
        train_batches = 0
        
        for batch_data in train_loader:
            if isinstance(batch_data, list):
                inputs = batch_data[0].to(device)
            else:
                inputs = batch_data.to(device)
            
            # Forward pass
            outputs = model(inputs)
            loss = criterion(outputs, inputs)
            
            # Backward pass
            optimizer.zero_grad()
            loss.backward()
            
            # Gradient clipping
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
            
            optimizer.step()
            
            train_loss += loss.item()
            train_batches += 1
        
        avg_train_loss = train_loss / train_batches
        train_losses.append(avg_train_loss)
        
        # Validation phase
        model.eval()
        val_loss = 0.0
        val_batches = 0
        
        with torch.no_grad():
            for batch_data in val_loader:
                if isinstance(batch_data, list):
                    inputs = batch_data[0].to(device)
                else:
                    inputs = batch_data.to(device)
                
                outputs = model(inputs)
                loss = criterion(outputs, inputs)
                
                val_loss += loss.item()
                val_batches += 1
        
        avg_val_loss = val_loss / val_batches
        val_losses.append(avg_val_loss)
        
        # Time
        epoch_time = time.time() - start_time
        
        # Print progress
        if (epoch + 1) % 5 == 0 or epoch == 0:
            print(f"{epoch+1:<10} {avg_train_loss:<15.6f} {avg_val_loss:<15.6f} {epoch_time:<10.2f}s")
        
        # Early stopping
        if avg_val_loss < best_val_loss:
            best_val_loss = avg_val_loss
            patience_counter = 0
            # Save best model
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'train_loss': avg_train_loss,
                'val_loss': avg_val_loss,
            }, f'outputs/models/{model_name}_best.pth')
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print(f"\nEarly stopping triggered at epoch {epoch + 1}")
                break
    
    print(f"\nTraining completed!")
    print(f"Best validation loss: {best_val_loss:.6f}")
    
    return train_losses, val_losses

def calculate_reconstruction_errors(model, data_loader, device):
    """
    Calculate reconstruction errors for all samples
    """
    model.eval()
    errors = []
    
    with torch.no_grad():
        for batch_data in data_loader:
            if isinstance(batch_data, list):
                inputs = batch_data[0].to(device)
            else:
                inputs = batch_data.to(device)
            
            outputs = model(inputs)
            
            # Calculate MSE for each sample
            batch_errors = torch.mean((inputs - outputs) ** 2, dim=(1, 2))
            errors.extend(batch_errors.cpu().numpy())
    
    return np.array(errors)

print("✓ Training functions defined!")
```

```python
# ============================================================================
# CELL 5: Create Data Loaders - Fraud Dataset
# ============================================================================
print("\n" + "="*80)
print("CREATING DATA LOADERS - FRAUD DATASET")
print("="*80)

BATCH_SIZE = 256

# Convert to PyTorch tensors
X_train_fraud_tensor = torch.FloatTensor(X_train_fraud)
X_val_fraud_tensor = torch.FloatTensor(X_val_fraud)

# Create datasets
train_dataset_fraud = TensorDataset(X_train_fraud_tensor)
val_dataset_fraud = TensorDataset(X_val_fraud_tensor)

# Create data loaders
train_loader_fraud = DataLoader(
    train_dataset_fraud,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=True if torch.cuda.is_available() else False
)

val_loader_fraud = DataLoader(
    val_dataset_fraud,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=True if torch.cuda.is_available() else False
)

print(f"Batch size: {BATCH_SIZE}")
print(f"Train batches: {len(train_loader_fraud)}")
print(f"Validation batches: {len(val_loader_fraud)}")
```

```python
# ============================================================================
# CELL 6: Initialize and Train LSTM-AE on Fraud Dataset
# ============================================================================
print("\n" + "="*80)
print("TRAINING LSTM-AUTOENCODER ON FRAUD DATASET")
print("="*80)

# Model hyperparameters
HIDDEN_DIM1 = 64
HIDDEN_DIM2 = 32
LATENT_DIM = 16
DROPOUT = 0.2
LEARNING_RATE = 0.001
NUM_EPOCHS = 100

print(f"Model hyperparameters:")
print(f"  Hidden dim 1: {HIDDEN_DIM1}")
print(f"  Hidden dim 2: {HIDDEN_DIM2}")
print(f"  Latent dim: {LATENT_DIM}")
print(f"  Dropout: {DROPOUT}")
print(f"  Learning rate: {LEARNING_RATE}")
print(f"  Max epochs: {NUM_EPOCHS}")

# Initialize model
model_fraud = LSTMAutoencoder(
    n_features=N_FEATURES_FRAUD,
    timestep=TIMESTEP,
    hidden_dim1=HIDDEN_DIM1,
    hidden_dim2=HIDDEN_DIM2,
    latent_dim=LATENT_DIM,
    dropout=DROPOUT
).to(device)

# Count parameters
total_params = sum(p.numel() for p in model_fraud.parameters())
trainable_params = sum(p.numel() for p in model_fraud.parameters() if p.requires_grad)
print(f"\nModel parameters:")
print(f"  Total: {total_params:,}")
print(f"  Trainable: {trainable_params:,}")

# Define loss and optimizer
criterion_fraud = nn.MSELoss()
optimizer_fraud = optim.Adam(model_fraud.parameters(), lr=LEARNING_RATE)

# Train the model
train_losses_fraud, val_losses_fraud = train_autoencoder(
    model=model_fraud,
    train_loader=train_loader_fraud,
    val_loader=val_loader_fraud,
    criterion=criterion_fraud,
    optimizer=optimizer_fraud,
    num_epochs=NUM_EPOCHS,
    device=device,
    model_name='lstm_ae_fraud'
)
```

```python
# ============================================================================
# CELL 7: Plot Training History - Fraud Dataset
# ============================================================================
print("\n" + "="*80)
print("PLOTTING TRAINING HISTORY - FRAUD DATASET")
print("="*80)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(train_losses_fraud, label='Training Loss', linewidth=2)
plt.plot(val_losses_fraud, label='Validation Loss', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('LSTM-AE Training History - Fraud Dataset')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(train_losses_fraud, label='Training Loss', linewidth=2)
plt.plot(val_losses_fraud, label='Validation Loss', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('LSTM-AE Training History - Fraud Dataset (Log Scale)')
plt.yscale('log')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('outputs/figures/08_lstm_ae_fraud_training_history.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# ============================================================================
# CELL 8: Load Best Model and Calculate Errors - Fraud
# ============================================================================
print("\n" + "="*80)
print("CALCULATING RECONSTRUCTION ERRORS - FRAUD DATASET")
print("="*80)

# Load best model
checkpoint = torch.load('outputs/models/lstm_ae_fraud_best.pth')
model_fraud.load_state_dict(checkpoint['model_state_dict'])
print(f"✓ Loaded best model from epoch {checkpoint['epoch'] + 1}")
print(f"  Best validation loss: {checkpoint['val_loss']:.6f}")

# Calculate reconstruction errors for validation set (normal only)
val_normal_indices = y_val_fraud == 0
X_val_fraud_normal = X_val_fraud[val_normal_indices]

val_normal_dataset = TensorDataset(torch.FloatTensor(X_val_fraud_normal))
val_normal_loader = DataLoader(val_normal_dataset, batch_size=BATCH_SIZE, shuffle=False)

errors_val_normal_fraud = calculate_reconstruction_errors(
    model_fraud, val_normal_loader, device
)

print(f"\nValidation normal errors shape: {errors_val_normal_fraud.shape}")
print(f"Error statistics:")
print(f"  Min: {errors_val_normal_fraud.min():.6f}")
print(f"  Max: {errors_val_normal_fraud.max():.6f}")
print(f"  Mean: {errors_val_normal_fraud.mean():.6f}")
print(f"  Std: {errors_val_normal_fraud.std():.6f}")
print(f"  Median: {np.median(errors_val_normal_fraud):.6f}")
```

```python
# ============================================================================
# CELL 10: Create Data Loaders - Olist Dataset
# ============================================================================
print("\n" + "="*80)
print("CREATING DATA LOADERS - OLIST DATASET")
print("="*80)

# Convert to PyTorch tensors
X_train_olist_tensor = torch.FloatTensor(X_train_olist)
X_val_olist_tensor = torch.FloatTensor(X_val_olist)

# Create datasets
train_dataset_olist = TensorDataset(X_train_olist_tensor)
val_dataset_olist = TensorDataset(X_val_olist_tensor)

# Create data loaders
train_loader_olist = DataLoader(
    train_dataset_olist,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=0,
    pin_memory=True if torch.cuda.is_available() else False
)

val_loader_olist = DataLoader(
    val_dataset_olist,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=True if torch.cuda.is_available() else False
)

print(f"Train batches: {len(train_loader_olist)}")
print(f"Validation batches: {len(val_loader_olist)}")
```

```python
# ============================================================================
# CELL 11: Initialize and Train LSTM-AE on Olist Dataset
# ============================================================================
print("\n" + "="*80)
print("TRAINING LSTM-AUTOENCODER ON OLIST DATASET")
print("="*80)

# Initialize model
model_olist = LSTMAutoencoder(
    n_features=N_FEATURES_OLIST,
    timestep=TIMESTEP,
    hidden_dim1=HIDDEN_DIM1,
    hidden_dim2=HIDDEN_DIM2,
    latent_dim=LATENT_DIM,
    dropout=DROPOUT
).to(device)

# Count parameters
total_params = sum(p.numel() for p in model_olist.parameters())
trainable_params = sum(p.numel() for p in model_olist.parameters() if p.requires_grad)
print(f"\nModel parameters:")
print(f"  Total: {total_params:,}")
print(f"  Trainable: {trainable_params:,}")

# Define loss and optimizer
criterion_olist = nn.MSELoss()
optimizer_olist = optim.Adam(model_olist.parameters(), lr=LEARNING_RATE)

# Train the model
train_losses_olist, val_losses_olist = train_autoencoder(
    model=model_olist,
    train_loader=train_loader_olist,
    val_loader=val_loader_olist,
    criterion=criterion_olist,
    optimizer=optimizer_olist,
    num_epochs=NUM_EPOCHS,
    device=device,
    model_name='lstm_ae_olist'
)
```

```python
# ============================================================================
# CELL 12: Plot Training History - Olist Dataset
# ============================================================================
print("\n" + "="*80)
print("PLOTTING TRAINING HISTORY - OLIST DATASET")
print("="*80)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(train_losses_olist, label='Training Loss', linewidth=2)
plt.plot(val_losses_olist, label='Validation Loss', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('LSTM-AE Training History - Olist Dataset')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.plot(train_losses_olist, label='Training Loss', linewidth=2)
plt.plot(val_losses_olist, label='Validation Loss', linewidth=2)
plt.xlabel('Epoch')
plt.ylabel('Loss (MSE)')
plt.title('LSTM-AE Training History - Olist Dataset (Log Scale)')
plt.yscale('log')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('outputs/figures/10_lstm_ae_olist_training_history.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# ============================================================================
# CELL 13: Load Best Model and Calculate Errors - Olist
# ============================================================================
print("\n" + "="*80)
print("CALCULATING RECONSTRUCTION ERRORS - OLIST DATASET")
print("="*80)

# Load best model
checkpoint = torch.load('outputs/models/lstm_ae_olist_best.pth')
model_olist.load_state_dict(checkpoint['model_state_dict'])
print(f"✓ Loaded best model from epoch {checkpoint['epoch'] + 1}")
print(f"  Best validation loss: {checkpoint['val_loss']:.6f}")

# Calculate reconstruction errors for validation set (normal only)
val_normal_indices_olist = y_val_olist == 0
X_val_olist_normal = X_val_olist[val_normal_indices_olist]

val_normal_dataset_olist = TensorDataset(torch.FloatTensor(X_val_olist_normal))
val_normal_loader_olist = DataLoader(val_normal_dataset_olist, batch_size=BATCH_SIZE, shuffle=False)

errors_val_normal_olist = calculate_reconstruction_errors(
    model_olist, val_normal_loader_olist, device
)

print(f"\nValidation normal errors shape: {errors_val_normal_olist.shape}")
print(f"Error statistics:")
print(f"  Min: {errors_val_normal_olist.min():.6f}")
print(f"  Max: {errors_val_normal_olist.max():.6f}")
print(f"  Mean: {errors_val_normal_olist.mean():.6f}")
print(f"  Std: {errors_val_normal_olist.std():.6f}")
print(f"  Median: {np.median(errors_val_normal_olist):.6f}")
```
